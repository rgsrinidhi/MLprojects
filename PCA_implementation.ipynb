```python
from os import EX_CONFIG
import numpy as np
X = np.array([[1,2,1],[3,4,3],[5,6,5],[7,8,7],[9,12,9]],dtype=float)
print(X)

#Step 1
#find x and y mean

mean = X.mean(axis = 0)
print(mean)
X_Centered = X-mean
print(X_Centered)
#Step 2
#Covariance matrix

cov_mat = np.cov(X_Centered, rowvar = False)
print(cov_mat)

#Step 3
#Eigen Values

eigvals, eigvecs = np.linalg.eigh(cov_mat)
print(eigvals)
# print(eigvecs)

#Step 4
# descendint order
order = np.argsort(eigvals)[::-1]
print("order")
print(order)
eigvals = eigvals[order]
eigvecs = eigvecs[:,order]

print(eigvals)
print(eigvecs)

#Step 5
#explained variance ratio

explained_variance_ratio = eigvals /eigvals.sum()
print(explained_variance_ratio)

#Step 6
X_pca = X_Centered @ eigvecs
print(X_pca)

```
